```python
class Karyawan:
    def __init__(self, nama, gaji):
        self.nama = nama
        self.gaji = gaji
    def cetak_kalimat(self):
        print(f' Karyawan {self.nama}, memiliki gaji {self.gaji}')
karyawan1=Karyawan("Siti", 5000000)
karyawan1.cetak_kalimat()
karyawan2=Karyawan("Izzi", 6000000)
karyawan2.cetak_kalimat()
karyawan3=Karyawan("Lala", 7000000)
karyawan3.cetak_kalimat()
karyawan4=Karyawan("Rani", 8000000)
karyawan4.cetak_kalimat()
karyawan5=Karyawan("Dinda",9000000)
karyawan5.cetak_kalimat()
```

```python
class Karyawan:
    def __init__(self, nama, gaji):
        self.nama = nama
        self.gaji = gaji
        self.energi=100
    def lembur(self):
        print(f"{self.nama} sedang lembur!")
        self.energi -=30
        self.gaji +=200000

data=[("Siti", 5000000), ("Izzi", 6000000), ("Lala", 7000000), ("Rani", 8000000), ("Dinda", 9000000),]
for nama, gaji in data:
    karyawan=Karyawan(nama, gaji)
    karyawan.lembur()
    print(f"sisa energi: {karyawan.energi}%")
    print(f"gaji terbaru: {karyawan.gaji}")
```

```python
class RekeningBank:
    def __init__(self, nama_pemilik, saldo=0):
        self.nama_pemilik = nama_pemilik
        self.saldo = saldo

    def menabung(self, nominal):
        self.saldo += nominal
        print(f"Berhasil menabung Rp{nominal}. Saldo sekarang: Rp{self.saldo}")

    def tarik_tunai(self, nominal):
        if self.saldo >= nominal:
            self.saldo -= nominal
            print(f"Berhasil menarik Rp{nominal}. Saldo sekarang: Rp{self.saldo}")
        else:
            print("Maaf, saldo tidak mencukupi.")
rekening_saya = RekeningBank("Hanifah N. A", 150000)
rekening_saya.tarik_tunai(50000)
rekening_saya.tarik_tunai(200000)
rekening_saya.menabung(300000)
```
